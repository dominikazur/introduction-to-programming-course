## 8. Implement the telemetry classifier

Use the four input variables below and an `if`/`elif`/`else` chain. Assign exactly one of these strings to `status`:

```text
CRITICAL
SAFE MODE
WARNING
NOMINAL
```

Do not start coding until your decision table is complete.

````python
battery_charge = 80
bus_voltage_v = 28
panel_temperature_c = 20
comms_ok = True

# Implement the priority rules here.
def spacecraft_health(battery_charge, bus_voltage_v, panel_temperature_c, comms_ok):
    if battery_charge < 15 or bus_voltage_v < 22:
        return "CRITICAL"

    elif (
        battery_charge < 25
        or bus_voltage_v < 24
        or panel_temperature_c < -20
        or panel_temperature_c > 85
        or not comms_ok
    ):
        return "SAFE MODE"

    elif battery_charge <= 35 or panel_temperature_c > 75:
        return "WARNING"

    else:
        return "NOMINAL"

print(spacecraft_health(battery_charge, bus_voltage_v, panel_temperature_c, comms_ok))
````

### Explain your implementation

Why must the CRITICAL test appear before the SAFE MODE test? Why is `else` appropriate for NOMINAL?

**Your explanation:**
It must be that way, since the SAFE MODE assumptions already include the NOMINAL ones (if battery charge is not below 25% it will not be below 15%, if bus voltage is not below 24V, it will not be below 22V)
